```python
# ==========================================
# ITERATION 1 — LOAD DATASET
# ==========================================

import pandas as pd

df = pd.read_csv('/content/BKB_WaterQualityData_2020084.csv')

print("Dataset Loaded Successfully")
```

```python
# ==========================================
# ITERATION 2 — INSPECT DATASET STRUCTURE
# ==========================================

print("Dataset Shape:")
print(df.shape)

print("\n")

print("Column Names:")
print(df.columns)

print("\n")

print("First 5 Rows:")
print(df.head())
```

```python
# ==========================================
# ITERATION 3 — CHECK DATA TYPES
# ==========================================

print(df.info())
```

```python
# ==========================================
# ITERATION 4 — CHECK MISSING VALUES
# ==========================================

print(df.isnull().sum())
```

```python
# ==========================================
# ITERATION 5 — CLEAN COLUMN NAMES
# ==========================================

df.columns = df.columns.str.strip()

print(df.columns)
```

```python
# ==========================================
# ITERATION 6 — INSPECT SAMPLE ROW VALUES
# ==========================================

print(df.iloc[0])

print("\n============================\n")

print(df.iloc[1])

print("\n============================\n")

print(df.iloc[2])
```

```python
# ==========================================
# ITERATION 7 — HANDLE MISSING VALUES
# ==========================================

df = df.fillna("Not Available")

print("Missing values handled successfully")
```

```python
# ==========================================
# ITERATION 8 — CREATE NATURAL LANGUAGE REPORTS
# ==========================================

documents = []

for _, row in df.iterrows():

    report = f"""

    Water quality report for site {row['Site_Id']}.

    Measurement date: {row['Read_Date']}.

    Salinity level recorded was {row['Salinity (ppt)']} ppt.

    Dissolved oxygen level measured {row['Dissolved Oxygen (mg/L)']} mg/L.

    Water pH value recorded was {row['pH (standard units)']}.

    Water temperature measured {row['Water Temp (?C)']} degrees Celsius.

    Water depth recorded was {row['Water Depth (m)']} meters.

    Air temperature measured {row['AirTemp (C)']} degrees Celsius.

    Secchi depth measured {row['Secchi Depth (m)']} meters.

    """

    documents.append(report)

print(documents[0])
```

```python
# ==========================================
# ITERATION 9 — CLEAN GENERATED TEXT
# ==========================================

import re

cleaned_docs = []

for doc in documents:

    doc = re.sub(r'\s+', ' ', doc)

    doc = doc.strip().lower()

    cleaned_docs.append(doc)

print(cleaned_docs[0])
```

```python
# ==========================================
# ITERATION 10 — TEXT CHUNKING
# ==========================================

!pip install langchain-text-splitters
```

```python
# ==========================================
# IMPORT TEXT SPLITTER
# ==========================================

from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
# ==========================================
# APPLY CHUNKING
# ==========================================

splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=50
)

chunks = splitter.create_documents(cleaned_docs)

print("Total Chunks Created:", len(chunks))
```

```python
# ==========================================
# ITERATION 11 — VERIFY CHUNK QUALITY
# ==========================================

print("Total Chunks Created:", len(chunks))

print("\n")

print("FIRST CHUNK:\n")

print(chunks[0].page_content)

print("\n=============================\n")

print("SECOND CHUNK:\n")

print(chunks[1].page_content)
```

```python
# ==========================================
# ITERATION 12 — INSTALL EMBEDDING LIBRARY
# ==========================================

!pip install sentence-transformers
```

```python
# ==========================================
# ITERATION 13 — LOAD EMBEDDING MODEL
# ==========================================

from sentence_transformers import SentenceTransformer

embedding_model = SentenceTransformer('all-MiniLM-L6-v2')

print("Embedding Model Loaded Successfully")
```

```python
# ==========================================
# ITERATION 14 — GENERATE EMBEDDINGS
# ==========================================

texts = [doc.page_content for doc in chunks]

embeddings = embedding_model.encode(texts)

print("Embedding Shape:")

print(embeddings.shape)
```

```python
# ==========================================
# ITERATION 15 — INSTALL FAISS
# ==========================================

!pip install faiss-cpu
```

```python
# ==========================================
# ITERATION 16 — IMPORT FAISS
# ==========================================

import faiss
import numpy as np

print("FAISS Imported Successfully")
```

```python
# ==========================================
# ITERATION 17 — CREATE VECTOR DATABASE
# ==========================================

dimension = embeddings.shape[1]

index = faiss.IndexFlatL2(dimension)

index.add(np.array(embeddings))

print("Total Vectors Stored:", index.ntotal)
```

```python
# ==========================================
# ITERATION 18 — USER QUERY
# ==========================================

query = "Which locations show poor water quality?"

print(query)
```

```python
# ==========================================
# ITERATION 19 — CONVERT QUERY INTO VECTOR
# ==========================================

query_embedding = embedding_model.encode([query])

print(query_embedding.shape)
```

```python
# ==========================================
# ITERATION 20 — RETRIEVE TOP MATCHES
# ==========================================

D, I = index.search(np.array(query_embedding), k=5)

print("Retrieved Indexes:")

print(I)
```

```python
# ==========================================
# ITERATION 21 — DISPLAY RETRIEVED CHUNKS
# ==========================================

retrieved_chunks = [texts[i] for i in I[0]]

for idx, chunk in enumerate(retrieved_chunks):

    print(f"\n================ CHUNK {idx+1} ================\n")

    print(chunk)
```

```python
# ==========================================
# ITERATION 22 — BUILD CONTEXT
# ==========================================

context = "\n".join(retrieved_chunks)

print(context)
```

```python
# ==========================================
# ITERATION 23 — CREATE FINAL PROMPT
# ==========================================

prompt = f"""
You are an expert AI system specialized in water management.

Use ONLY the provided context to answer the question.

If the answer is unavailable in the context, reply:
'Information not available in the provided data.'

Context:
{context}

Question:
{query}
"""

print(prompt)
```
